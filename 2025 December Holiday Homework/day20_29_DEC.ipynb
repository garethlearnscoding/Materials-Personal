## Task 1.1
Write program code to declare all the classes described above.
<div style="text-align: right">[18]</div>



````python
# Task 1.1

class Animal:
    def __init__(self,name,mass,volume,age):
        self.__name = name
        self.__mass = int(mass)
        self.__volume = int(volume)
        self.__age = float(age)
    def get_name(self):
        return self.__name
    def get_mass(self):
        return self.__mass
    def get_volume(self):
        return self.__volume
    def get_age(self):
        return self.__age
    def isSick(self):
        return True if self.get_mass()/self.get_volume() < 0.8 else False
    def __str__(self):
        return f"Name: {self.get_name()} Age: {self.get_age()} Mass: {self.get_mass()} Volume: {self.get_volume()}"

class WarmBloodAnimal(Animal):
    def __init__(self,name,mass,volume,age,bodyTemp):
        super().__init__(name,mass,volume,age)
        self.bodyTemp = float(bodyTemp)

class ColdBloodAnimal(Animal):
    def __init__(self,name,mass,volume,age,isPoisonous):
        super().__init__(name,mass,volume,age)
        self.__isPoisonous = bool(isPoisonous)
    def getIsPoisonous(self):
        return self.__isPoisonous

class Dog(WarmBloodAnimal):
    def __init__(self,name,mass,volume,age,bodyTemp,hasOwner):
        super().__init__(name,mass,volume,age,bodyTemp)
        self.hasOwner = hasOwner
    def hasFever(self):
        return True if self.bodyTemp > 39.0 else False
    def isSick(self):
        return True if super().isSick() or self.hasFever() else False
    def isYoung(self):
        return True if self.get_age() < 1.0 else False
    def __str__(self):
        inherited = super().__str__()
        sickness = " Sick" if self.isSick() else " Well"
        youth = " Young" if self.isYoung() else " Adult"
        return inherited + sickness + youth
    
class Goose(WarmBloodAnimal):
    def __init__(self,name,mass,volume,age,bodyTemp):
        super().__init__(name,mass,volume,age,bodyTemp)
    def hasFever(self):
        return True if self.bodyTemp > 43.0 else False
    def isSick(self):
        return True if super().isSick() or self.hasFever() else False
    def isYoung(self):
        return True if self.get_age()<0.5 else False
    def __str__(self):
        inherited = super().__str__()
        sickness = " Sick" if self.isSick() else " Well"
        youth = " Young" if self.isYoung() else " Adult"
        return inherited + sickness + youth

class Frog(ColdBloodAnimal):
    def __init__(self,name,mass,volume,age,isPoisonous):
        super().__init__(name,mass,volume,age,isPoisonous)
    def isYoung(self):
        return True if self.get_age() < 0.3 else False
    def __str__(self):
        inherited = super().__str__()
        sickness = " Sick" if self.isSick() else " Well"
        youth = " Young" if self.isYoung() else " Adult"
        return inherited + sickness + youth

class Snake(ColdBloodAnimal):
    def __init__(self,name,mass,volume,age,isPoisonous,length):
        super().__init__(name,mass,volume,age,isPoisonous)
        self.length = float(length)
    def isSick(self):
        return True if self.get_mass() < 0.8*(0.00035*(self.length*100)**3) else False
    def isYoung(self):
        return True if self.get_age() < 0.5 else False
    def __str__(self):
        inherited = super().__str__()
        sickness = " Sick" if self.isSick() else " Well"
        youth = " Young" if self.isYoung() else " Adult"
        return inherited + sickness + youth
````

## Task 1.3
Write a function, `task1_3()` to:
- read the file `“Animals_lst.txt”`
- process its content and create the appropriate instances for each animal
- add all the instances in a Python list and return it.
The format of the information in `“Animals_lst.txt”` is as follow:
- If it is a dog,
    ```
    Dog, name, mass, volume, age, body_temperature, hasOwner
    ```
- If it is a goose,
    ```
    Goose, name, mass, volume, age, body_temperature
    ```
- If it is a frog,
    ```
    Frog, name, mass, volume, age, isPoisonous
    ```
- If it is a snake,
    ```
    Snake, name, mass, volume, age, isPoisonous, length
    ```
<div style="text-align: right">[3]</div>


````python
import csv

def task1_3():
    classes = {
        "Dog": Dog,
        "Goose": Goose,
        "Frog":Frog,
        "Snake":Snake,
    }

    with open("./resources/Animals_lst.txt") as f:
        data = [
            classes[row[0]](*row[1:])
            for row in csv.reader(f)
        ]
    return data
````

## Task 1.5
Write a function, `task1_5(a_list, animal_type)` to:
- find all the young animal of animal_type that has its volume larger than that of the smallest adult animal of the same animal type
- return the name of this group of young animals in a list.
<div style="text-align: right">[4]</div>


````python
def task1_5(a_list,animal_type):
    instances = {"Young":[],"Adult":0}
    classes = {
        "Dog": Dog,
        "Goose": Goose,
        "Frog":Frog,
        "Snake":Snake,
    }
    for i in a_list:
        if isinstance(i,classes[animal_type]):
            if not i.isYoung():
                if i.get_volume > instances["Adult"]:
                    instances["Adult"] = i.get_volume()
    for i in a_list:
        if isinstance(i,classes[animal_type]):
            if i.isYoung():
                if i.get_volume() < instances["Adult"]:
                    instances["Young"] += [i.get_name()]
    return instances["Young"]
````

## Task 1.6
Write a function, `task1_6(a_list)` to:
- filter out all the Cold Blood Animal in `a_list`

- return only the names of the cold blood animals in a list sorted according 
to their mass in ascending order.
<div style="text-align: right">[4]</div>

Test your program with the following test data:

<div style="text-align: right">[1]</div>


````python
a_list = task1_3()

def task1_6(a_list):
    instances = []
    for i in a_list:
        if isinstance(i,ColdBloodAnimal):
            instances.append(i)
    
    def merge(a,b):
        c = []
        while len(a) >= 1 and len(b) >= 1:
            if a[0].get_mass() > b[0].get_mass():
                c.append(b.pop(0))
            else:
                c.append(a.pop(0))
        c += a if len(a) else b
        return c

    def mergesort(array):
        mid = len(array)//2
        if len(array) <= 1:
            return array

        left = mergesort(array[:mid])
        right = mergesort(array[mid:])
        
        return merge(left,right)

    sorted = mergesort(instances)
    return [i.get_name() for i in sorted]

task1_6(a_list)
````
