Write program code to:

- implement a bubble sort to sort the unsorted 1-dimensional list from Task 2.1 (you must not use a built-in function).

Write a program to:

<div style="text-align: right">[5]</div>

<div style="text-align: right">[1]</div> 

```python
# Task 2.2


def bubble(a):
    for i in range(len(a)):
        swap = False
        for j in range(len(a)-i-1):
            if a[j]>a[j+1]:
                a[j],a[j+1] = a[j+1],a[j]
                swap = True
        if not swap:
            break
    return a

task2_1()
bubble(lst)
print(lst)
```

```python
# Task 2.3

def merge(a,b):
    c = []
    while len(a) >= 1 and len(b) >= 1:
        if a[0] > b[0]:
            c.append(b.pop(0))
        else:
            c.append(a.pop(0))
    if len(a):
        c+= a
    else:
        c+=b
    return c

def mergesort(array):
    mid = len(array)//2
    if len(array) <= 1:
        return array
    
    a = mergesort(array[:mid])
    b = mergesort(array[mid:])

    return merge(a,b)


while True:
    action = input("Enter Sorting Algorithm: ")
    if action == "Merge":
        task2_1()
        mergesort(lst)
    elif action == "Bubble":
        task2_1()
        bubble(lst)
    else:
        continue
    print(lst)
    break
```
