## Task 2
The file `DATA.txt` contains the comma-separated values to be stored in the database.

Write a Python program to read the data from the file and then store it in the database.

Save your Python program as
`Task4_2_<your name>_<centre number>_<index number>.py`

<div style="text-align: right">[6]</div>

```python
# Task 2
import csv
import sqlite3

with open("./resources/DATA.txt") as f:
    data = list(csv.reader(f))

members_data = {(i[0],i[1],i[2]) for i in data}
competitions_data = {(i[4],i[3]) for i in data}
scores_data = [[i[0],i[-2],i[-1]] for i in data]

members_query = "INSERT INTO Members(Mem_ID,Mem_Name,Mem_DOB) VALUES (?,?,?)"
competitions_query = "INSERT INTO Competitions(Comp_ID,Comp_Name) VALUES (?,?)"
scores_query = "INSERT INTO Scores(Mem_ID,Comp_ID,Comp_Score) VALUES(?,?,?)"

conn = sqlite3.connect("./resources/club.db")
conn.executemany(members_query,members_data)
conn.executemany(competitions_query,competitions_data)
conn.executemany(scores_query,scores_data)
conn.commit()
conn.close()
```
