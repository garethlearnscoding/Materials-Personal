## Task 2.2 
The computer randomly generates the position of a battleship within the grid. It will occupy 4 units in length, represented by 4 consecutive `S` either horizontally or vertically within the grid. 

The computer will also generate 7 random locations to represent the rocks (`R`). 

Write program code to: 
- generate 4 `(x,y)` coordinates to represent the battleship (`S`) within the grid and 
store them in a suitable data structure. 
- generate 7 `(x,y)` coordinates to represent the position of the rocks (`R`) within the 
grid and store them in a suitable data structure. 
- display the board showing the battleship and the rocks. 

A possible board is as follows: 

<center>

```
\ABCDEFGHIJ 
1........R. 
2..SSSS.... 
3...R...... 
4.....R.... 
5.R.....R.. 
6...R...... 
7.......R.. 
``` 

</center>
<div style="text-align: right">[6]</div>

````python
# Task 2.2 

import random

board = [list("\\ABCDEFGHIJ")]+[[str(i)]+["."]*10 for i in range(1,8)]

ship_y = [random.randint(1,7)]*4
first_x = random.randint(1,7)
ship_x = [first_x + i for i in range(4)]
ship_cords = list(zip(ship_x,ship_y))
print(ship_cords)
rocks = []

while len(rocks) != 7:
    xy = (random.randint(1,10),random.randint(1,7))
    if xy not in ship_cords:
        rocks.append(xy)

print(rocks)
    
for i in ship_cords:
    board[i[1]][i[0]] = "S"

for i in rocks:
    board[i[1]][i[0]] = "R"


for i in board:
    print("".join(i))
````


## Task 2.3 
During the game, the position of the battleship will not be revealed. The location of the battleship will be represented by `‘.’`. 

The player will fire a missile to strike at a target location. If the missile hits the battleship, it will be indicated with a `‘H’`; Otherwise, it will be indicated with a `‘M’` for a miss. 

The game ends when player has hit all the 4 `(x,y)` coordinates representing the 
battleship. 

Write program code to: 
- display the board with the rocks (`‘R’`) and hide the position of the battleship with '.'. 
- prompt the player to enter the coordinates to target the missile. 
    For example, 
    
    ```
    Enter missile coordinate (e.g. B4):
    ```
- display the board indicating a hit or miss with `‘H'` or `‘M‘`. 
- allow the player to continue the game until all the 4 `(x,y)` coordinates of the 
battleship have been hit. 
- display the number of missiles fired at the end of the game. 

<div style="text-align: right">[10]</div>

````python
# Task 2.2 

import random

board = [list("\\ABCDEFGHIJ")]+[[str(i)]+["."]*10 for i in range(1,8)]

ship_y = [random.randint(1,7)]*4
first_x = random.randint(1,7)
ship_x = [first_x + i for i in range(4)]
ship_cords = list(zip(ship_y,ship_x))
print(ship_cords)
rocks = []

while len(rocks) < 7:
    xy = (random.randint(1,7),random.randint(1,10))
    if xy not in ship_cords:
        rocks.append(xy)

for i in rocks:
    board[i[0]][i[1]] = "R"

# for i in board:
#     print("".join(i))

hit = 0
col_dict = {"A":1,"B":2,"C":3,"D":4,"E":5,"F":6,"G":7,"H":8,"I":9,"J":10}
while hit < 4:
    for i in board:
        print("".join(i))
    missile = list(input("Enter missile coordinate (e.g. B4): "))
    print(missile)
    missile_coord = (int(missile[1]),col_dict[missile[0]])
    if missile_coord in ship_cords:
        symbol = "H"
        hit += 1
    else:
        symbol = "M"
    board[missile_coord[0]][missile_coord[1]] = symbol


````
