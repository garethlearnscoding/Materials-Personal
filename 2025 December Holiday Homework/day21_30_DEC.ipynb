You are tasked with analysing a collection of bakery recipes. The recipe data is stored in a file named `RECIPES.TXT`.

For each recipe, the name of the bakery item will be listed first, followed by information about each ingredient for that item separated by commas:
```
<ingredient name>,<quantity required>,<unit of measurement>
```

For example, the first recipe in the file is for the Cheesecake item:
>```
> Cheesecake
> cream cheese,450,g
> sugar,150,g
> eggs,3,pieces
> sour cream,120,ml
> vanilla extract,1,tsp
> flour,30,g
>```

The recipes are separated by blank lines in the file `RECIPES.TXT`.

## Task 1.1
Write program code for a function named `task1_1` that takes the filename as a parameter and finds the most common ingredient and returns the name of the most common ingredient in the recipes, along with the number of occurrences.
<div style="text-align: right">[5]</div>
Run your program and show the output.	
<div style="text-align: right">[1]</div>

````python
# Task 1.1
import csv

def task1_1(filename):
    with open(f"./resources/{filename}") as f:
        data = list(csv.reader(f))
    ingredients = [i[0] for i in data if len(i) == 3]
    counter = {}
    for i in set(ingredients):
        counter[i] = ingredients.count(i)
    most_common = sorted(enumerate(counter))[-1][1]
    return most_common


print(task1_1("RECIPES.TXT"))
````

## Task 1.2
Write program code for an insertion sort function named `task1_2` that takes a list of strings, sorts it in ascending alphabetical order, and returns the sorted list.	
<div style="text-align: right">[4]</div>

````python
def task1_2(array):
    for i in range(1,len(array)):
        while i > 0 and array[i]<array[i-1]:
            array[i],array[i-1] = array[i-1],array[i]
            i -= 1
    return array
````

## Task 1.3
Write program code for a function named `task1_3` that reads the data from the file and calculates the total quantity of each ingredient required across all recipes. If the unit of measurement for an ingredient is the same in multiple recipes, sum up the quantities. Sort the names of ingredients alphabetically using the insertion sort function from Task 1.2 and display the sorted list of ingredients, along with their total quantities and units.
<div style="text-align: right">[6]</div>
Run your program and show the output.
<div style="text-align: right">[1]</div>

````python
import csv

def task1_3(): 
    with open("./resources/RECIPES.TXT") as f:
        data = [i for i in csv.reader(f) if len(i) == 3]
    cook_dict = {}
    for i in data:
        if i[0] not in cook_dict.keys():
            cook_dict[i[0]] = [i]
        else:
            cook_dict[i[0]] = cook_dict[i[0]]+[i]
    ingredients_dict = {}
    for ingredient,quants in cook_dict.items():
        quants_dict = {}
        for i in quants:
            if i[-1] not in quants_dict.keys():
                quants_dict[i[-1]] = [float(i[1])]
            else:
                quants_dict[i[-1]] = quants_dict[i[-1]]+[float(i[1])]
        ingredients_dict[ingredient] = quants_dict
    quants_dict = {}
    for ingredient,quants in ingredients_dict.items():
        ingredient_list = []
        for unit,measurements in quants.items():
            temp = [ingredient,sum(measurements),unit]
            ingredient_list.append(temp)
        quants_dict[ingredient] = ingredient_list
    sorted_ingredients = task1_2(list(quants_dict.keys()))
    for k in sorted_ingredients:
        for i in quants_dict[k]:
            print(i)
        
task1_3()
````

````python
import csv

def task1_3():
    totals = {}  # ingredient -> {unit -> total}

    with open("./resources/RECIPES.TXT") as f:
        for row in csv.reader(f):
            if len(row) != 3:
                continue
            ingredient, quantity, unit = row
            quantity = float(quantity)

            if ingredient not in totals:
                totals[ingredient] = {}
            if unit not in totals[ingredient]:
                totals[ingredient][unit] = 0.0

            totals[ingredient][unit] += quantity

    # Sort ingredients using task1_2
    sorted_ingredients = task1_2(list(totals.keys()))

    # Print results
    for ingredient in sorted_ingredients:
        for unit, total in totals[ingredient].items():
            print([ingredient, total, unit])

task1_3()
````
