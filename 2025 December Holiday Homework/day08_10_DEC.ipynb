## Task 3.1
Write a function `inputPuzzle()` to allow the user to input a number into each of the boxes by:
- taking 9 rows of 9 digits as input from the' user with a space between each digit, for example for one row: 1 2 3 4 5 6 7 8 9
- continually requesting a new row until the row is valid. A row is valid if it contains 9 digits that are all between 1 and 9 (inclusive). A row can have two or more of the same digits
- splitting each row into a list of digits
- appending each list of digits to a 2-dimensional list
- returning the 2-dimensional list.

<div style="text-align: right">[7]</div>

Test your program by entering the data for the puzzles shown in Example 1 and Example 2. Store the puzzles in variables named `example1` and `example2` respectively. 

<div style="text-align: right">[1]</div> 

```python
# Task 3.1

def inputPuzzle():
    puzzle = []
    for i in range(9):
        while True:
            cur_row = input(f"Row {i+1} | Enter answer: ")
            cur_row = [int(i) for i in cur_row.split(" ")]
            print(len(cur_row))
            print(set(cur_row))
            if len(cur_row) != 9:
                continue
            print(set(cur_row).issubset({1,2,3,4,5,6,7,8,9}))
            if set(cur_row).issubset({1,2,3,4,5,6,7,8,9}):
                break
        puzzle.append(cur_row)

example1 = inputPuzzle()
example2 = inputPuzzle()
```

## Task 3.3
Write program code to declare a function that checks each column of the puzzle to make sure it is complete vertically (contains each of the numbers 1 to 9 only once). The function should return -1 if the puzzle is complete vertically, or the column number where there is a duplicate. 
<div style="text-align: right">[3]</div> 

Test your program with both of the puzzles you stored in Task 3.1 by calling the function with each puzzle and outputting an appropriate message depending on the return value. 
<div style="text-align: right">[3]</div> 

```python
def vert_check(puzzle):
    transposed = list(zip(*puzzle))
    return hor_check(transposed)

print(vert_check(example1))
print(vert_check(example2))
```

Write program code to declare a function that checks each of the 9 mini-squares to make sure they are all complete (contain each of the numbers 1 to 9 only once). The function should return `True` if all mini-squares are complete, or `False` if any mini-squares are **not** complete.

```python
# Task 3.4





def ms_check(puzzle):
    mini_squares = [block[i]+block[i+1]+block[i+2] for i in range(0,9,3) for block in list(zip(*[[row[0:3],row[3:6],row[6:9]] for row in puzzle]))]
    return hor_check(mini_squares)

print(ms_check(example1))
print(ms_check(example2))
```
