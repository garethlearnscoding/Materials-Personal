**Sieve of Eratosthenes**

To find the prime numbers up to a certain number, n : 

An array is created to store all numbers up to n. 

If a number is prime, boolvalue is set to True while all boolvalues of the  multiples of that number up to n is set to False.


[Video Explaining the algorithms above](https://www.youtube.com/watch?v=fwxjMKBMR7s)
[Wikipedia Page](https://en.wikipedia.org/wiki/Sieve_of_Eratosthenes)

````python
from functools import lru_cache
# >> Sieve of Eratosthenes 
@lru_cache(None)
def is_prime_SoE(n):
    prime_list = []
    root = n**0.5 + 2
    if root%1 == 0:
        return False
    int_list = [True]*int(root)
    int_list[0] = int_list[1] = False
    for fac,boolval in enumerate(int_list):
        if boolval == True:
            if n%fac == 0:
                return False
            for multiple in range(fac*fac,int(root),fac):  
                int_list[multiple]=[False]        
    return True


# is_prime_SoE(112345678987656543)
is_prime_SoE(9999999999999999999)
````

**Dijkstra's Hidden Prime Finding Algorithm**

To find the prime numbers up to a certain number, n:

A pool is used to track past prime factors and their multiples, where it contains tuples (factor, multiple). As the divisor is increased, if it is larger then the lowest multiple in all tuples in said pool, it is not prime and the lowest multiple is also increased by its own factor. 


[Video Explaining the algorithms above](https://www.youtube.com/watch?v=fwxjMKBMR7s)


````python
def is_prime_dijkstra(n):
    pool = []
    primes_list=[2]
    if type(root := n**0.5 + 2) != float:
        return False
    i = 2
    pool.append([i,i*i])
    for i in range(3,n):
        print(f"I: {i}")
        pool.sort(key= lambda x:x[1])
        print(f"Pool: {pool}")
        min_factor_set = pool[0]
        print(f"Min Factor Set: {min_factor_set}")
        if i == min_factor_set[1]:
            multiple = min_factor_set.pop(1)
            multiple += min_factor_set[0]
            min_factor_set.insert(1,multiple)
            print(f"Min Factor Set: {min_factor_set}")
        elif i < min_factor_set[1]:
            pool.sort(key= lambda x:x[0])
            print(f"Pool: {pool}")
            primes_list.append(i)
            pool.append([i,i*i])
        else:
            pass
    return primes_list

is_prime_dijkstra(25)
````

## Exercise 4.19 Euclidean Algorithm

A more efficient method to compute the greatest common divisor of two integers $a$ and $b$ is by using the Euclidean algorithm. It is based on the principle that the greatest common divisor of two numbers does not change if the larger number is replaced by its difference with the smaller number. For example,

$$
\begin{align*}
\gcd(252,105)&=21\\ 
&=\gcd(252-2(105)=42,105)\\
&=\gcd(42,105-2(42)=21)\\
&=\gcd(42,21)
\end{align*}
$$

In particular, assume $a>b$, then by Division Algorithm, there exists $q\in \mathbb{Z}$ such that $a=qb+r$, where $0\leq r<b$. Then, 
$$
\begin{align*}
\gcd(a,b)&=\gcd(a-qb,b)\\
&=\gcd(r,b).
\end{align*}
$$
