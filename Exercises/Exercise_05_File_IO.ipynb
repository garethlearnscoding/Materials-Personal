- `'Parse%'`  refers to the performance of the character relative to all other players in the world,

- `'iLvL%'` refers to the performance of the player relative to players with similar quality of equipment,
- `'Active Time'` refers to the percentage of time spend by the character doing damage during the fight.
- `'DPS'` refers to the amount of damage done per second in the fight.

- Store the contents of the csv file into a list `records` where each element in the list is itself a list where the elements correspond to the `Parse%`,`Name`,`Amount`,`iLvL`,`iLvL%`,`Active Time`,`DPS` values for each line in the csv file.

## Exercise 5.9

A text file, `NOI_24_RESULT.txt` contains the results of the medallist recipients for the 24th National Olympiad in Informatics.

Write a code to:

- read the data from the text file,
- store the school name and medal counts in a Python dictionary `medal_count` where the school name is the key and the associated value is a tuple of the form `(GOLD, SILVER, BRONZE)`,
- print out the top 5 school with the highest number of medals won, regardless of the medal type. 

````python
import csv
medals = ["Gold","Silver","Bronze"]
medal_count = {}
with open("resources/NOI_24_RESULT.csv") as noi:
    reader = csv.reader(noi)
    for i in reader:
        try:
            # print(i[-1])
            school = i[-1].strip(" ")
            no_winners = len(i) - 1
            if school in medals:
                checker = medals.index(school)
            elif school not in medal_count.keys():
                medal_count[school] = [0,0,0]
            list_medals = medal_count.get(school)
            list_medals[checker] += no_winners
        except:
            pass

for k in medal_count.keys():
    lst = medal_count[k]
    tup = tuple(lst)
    summation = sum(tup)
    medal_count[k] = [tup,summation]

for k,v in list(medal_count.items()):
    # print(f"{str(k):<55} {str(v[0]):>10}")
    itemsss = list(medal_count.items())
    itemsss.sort(reverse=True,key = lambda x: x[1][1])

print([i[0] for i in itemsss[0:5]])

````

### Task 2

Suppose you want to play the market by buying and selling the 6 items above, one way to decide on items to invest in is by using the mean and variance of the daily rate of returns of the items. The latter is used to quantify the risk of holding such item.

For an item $j$, 
- the **daily rate of return** of day $i$ to $i+1$, denoted as $r_i$ is defined to be $$r_i=\frac{p_{i+1}-p_{i}}{p_i},$$ where $p_{i+1}$ and $p_i$ are prices of the item on day $i+1$ and day $i$ respectively,
- the mean of the daily rate of returns, denoted as $\overline{r}$, is just $\frac{\sum r_i}{n}$, where $n$ is the number of daily rate of returns recorded,
- the variance of the daily rate of returns is $$\frac{\sum{(r_i-\overline{r})^2}}{n},$$ where $\overline{r}$ is the mean of the daily rate of returns and again $n$ is the number of daily returns recorded.

Write a program to:

- for each of the 6 items:
    - find the mean of the daily rate of return,
    - find the variance of the daily rate of return.

````python
# Task 1
import csv
widowbloom,widowbloom_pricemax,widowbloom_pricemin = [],[],[]
rising_glory,rising_glory_pricemax,rising_glory_pricemin = [],[],[]
marrowroot,marrowroot_pricemax,marrowroot_pricemin = [],[],[]
vigils_torch,vigils_torch_pricemax,vigils_torch_pricemin = [],[],[]
nightshade,nightshade_pricemax,nightshade_pricemin = [],[],[]
spectral_flask,spectral_flask_pricemax,spectral_flask_pricemin = [],[],[]

with open("resources/market_prices.csv") as market_prices:
    reader = csv.DictReader(market_prices)
    for i in reader:
        name = i["name_enus"]
        # print(name)
        if name == "Widowbloom":
            widowbloom.append(i)
            widowbloom_pricemax.append(int(i["pricemax"]))
            widowbloom_pricemin.append(int(i["pricemin"]))
        if name == "Rising Glory":
            rising_glory.append(i)
            rising_glory_pricemax.append(int(i["pricemax"]))
            rising_glory_pricemin.append(int(i["pricemin"]))
        if name == "Marrowroot":
            marrowroot.append(i)
            marrowroot_pricemax.append(int(i["pricemax"]))
            marrowroot_pricemin.append(int(i["pricemin"]))
        if name == "Vigil's Torch":
            vigils_torch.append(i)
            vigils_torch_pricemax.append(int(i["pricemax"]))
            vigils_torch_pricemin.append(int(i["pricemin"]))
        if name == "Spectral Flask of Power":
            spectral_flask.append(i)
            spectral_flask_pricemax.append(int(i["pricemax"]))
            spectral_flask_pricemin.append(int(i["pricemin"]))
        if name == "Nightshade":
            nightshade.append(i)
            nightshade_pricemax.append(int(i["pricemax"]))
            nightshade_pricemin.append(int(i["pricemin"]))


# >> Task 1
def task_1():
    print(f"Max Price{"."*60}")
    print(f"{"Widowbloom:":<55} {max(widowbloom_pricemax):>10}")
    print(f"{"Rising Glory:":<55} {max(rising_glory_pricemax):>10}")
    print(f"{"Marrowroot:":<55} {max(marrowroot_pricemax):>10}")
    print(f"{"Vigil's Torch:":<55} {max(vigils_torch_pricemax):>10}")
    print(f"{"Spectral Flask of Power:":<55} {max(spectral_flask_pricemax):>10}")
    print(f"{"Nightshade:":<55} {max(nightshade_pricemax):>10}")

    print(f"\nMin Price{"."*60}")
    print(f"{"Widowbloom:":<55} {min(widowbloom_pricemin):>10}")
    print(f"{"Rising Glory:":<55} {min(rising_glory_pricemin):>10}")
    print(f"{"Marrowroot:":<55} {min(marrowroot_pricemin):>10}")
    print(f"{"Vigil's Torch:":<55} {min(vigils_torch_pricemin):>10}")
    print(f"{"Spectral Flask of Power:":<55} {min(spectral_flask_pricemin):>10}")
    print(f"{"Nightshade:":<55} {min(nightshade_pricemin):>10}")
    print("\n")


# >> Task 2

def daily_ror(a,b):
    d1 = a.get("priceavg")
    d2 = b.get("priceavg")
    r = int(d2)/int(d1) - 1
    return r

def average(l):
    return sum(l)/len(l)


widowbloom_ror = list(map(daily_ror,widowbloom,widowbloom[1::]))
widowbloom_ror_avg = average(widowbloom_ror)
rising_glory_ror = list(map(daily_ror,rising_glory,rising_glory[1::]))
rising_glory_ror_avg = average(rising_glory_ror)
marrowroot_ror = list(map(daily_ror,marrowroot,marrowroot[1::]))
marrowroot_ror_avg = average(marrowroot_ror)
vigils_torch_ror = list(map(daily_ror,vigils_torch,vigils_torch[1::]))
vigils_torch_ror_avg = average(vigils_torch_ror)
spectral_flask_ror = list(map(daily_ror,spectral_flask,spectral_flask[1::]))
spectral_flask_ror_avg = average(spectral_flask_ror)
nightshade_ror = list(map(daily_ror,nightshade,nightshade[1::]))
nightshade_ror_avg = average(nightshade_ror)


widowbloom_ror_daily_variance = list(map(lambda x: (x-widowbloom_ror_avg)**2,widowbloom_ror))
widowbloom_variance = average(widowbloom_ror_daily_variance)
rising_glory_ror_daily_variance = list(map(lambda x: (x-rising_glory_ror_avg)**2,rising_glory_ror))
rising_glory_variance = average(rising_glory_ror_daily_variance)
marrowroot_ror_daily_variance = list(map(lambda x: (x-marrowroot_ror_avg)**2,marrowroot_ror))
marrowroot_variance = average(marrowroot_ror_daily_variance)
vigils_torch_daily_variance = list(map(lambda x: (x-vigils_torch_ror_avg)**2,vigils_torch_ror))
vigils_torch_variance = average(vigils_torch_daily_variance)
spectral_flask_ror_daily_variance = list(map(lambda x: (x-spectral_flask_ror_avg)**2,spectral_flask_ror))
spectral_flask_variance = average(spectral_flask_ror_daily_variance)
nightshade_ror_daily_variance = list(map(lambda x: (x-nightshade_ror_avg)**2,nightshade_ror))
nightshade_variance = average(nightshade_ror_daily_variance)


def task_2():
    print(f"Mean Daily Rate of Return{"."*44}")
    print(f"{"Widowbloom:":<55} {widowbloom_ror_avg:>10.05f}")
    print(f"{"Rising Glory:":<55} {rising_glory_ror_avg:>10.05f}")
    print(f"{"Marrowroot:":<55} {marrowroot_ror_avg:>10.05f}")
    print(f"{"Vigil's Torch:":<55} {vigils_torch_ror_avg:>10.05f}")
    print(f"{"Spectral Flask of Power:":<55} {spectral_flask_ror_avg:>10.05f}")
    print(f"{"Nightshade:":<55} {nightshade_ror_avg:>10.05f}")

    print(f"Variance in Daily Rate of Return{"."*37}")
    print(f"{"Widowbloom:":<55} {widowbloom_variance:>10.05f}")
    print(f"{"Rising Glory:":<55} {rising_glory_variance:>10.05f}")
    print(f"{"Marrowroot:":<55} {marrowroot_variance:>10.05f}")
    print(f"{"Vigil's Torch:":<55} {vigils_torch_variance:>10.05f}")
    print(f"{"Spectral Flask of Power:":<55} {spectral_flask_variance:>10.05f}")
    print(f"{"Nightshade:":<55} {nightshade_variance:>10.05f}")


def main():
    task_1()
    task_2()


main()
````

The $\operatorname{atan2}(y, x)$ function is used to find the angle between the positive $x$-axis and the ray from the origin $(0, 0)$ to $(x, y)$.

</center>

Take note that $\operatorname{atan2}()$ takes 2 arguments while $\operatorname{atan}()$ takes only 1 argument

Here is a table to illustrate the differences.

<center>

|Points  |$\operatorname{atan}(\frac{y}{x})$  |$\operatorname{atan2}(y, x)$|
|------  |-----------    |-----------|
|$(1, 1)$  |$\frac\pi4$    |$\frac\pi4$|
|$(-1, -1)$|$-\frac{3\pi}4$|$\frac\pi4$|

|$(1, 0)$  |$\pi$          |$0$        |
